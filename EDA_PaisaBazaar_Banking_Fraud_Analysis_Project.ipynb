## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
# let us first drop columns that are unnecessary or might contain sensitive information
df.drop(['ID','Name','SSN'],axis=1,inplace=True, errors='ignore')
#we saw that the age column had a float dtype instead of int so we fix that:
df['Age']=df['Age'].astype(int)
# now let's check for any uncertainities or messy data present in categorical columns that will be used in analysis
for col in df.select_dtypes(include='object'):
  print(col)
  print(df[col].unique())
```

```python
#So from the above categorical columns one of the useful column 'Payment_of_Min_Amount' has a Typing error where certain rows have No as NM
df['Payment_of_Min_Amount']=df['Payment_of_Min_Amount'].replace('NM','No')
#check
df['Payment_of_Min_Amount'].unique()
```

### What all manipulations have you done and insights you found?

1)During the info inspection we found out that the age column had a datatype of 'float' which was incorrect so we changed it to 'int' datatype<br>
2)We then dropped the columns like name, ID and SSN because they were irrelevant from an analysis perspective and also because these are sensitive data<br>
3)While checking for uncertainities in categorical columns using the unique method we found that the column 'Payment_of_Min_Amount' had large number of rows with a typing error where 'No' was written as 'NM' so we fixed that.<br>
4)In numerical columns certain columns seemed full of outliers like Annual_Income, Monthly_Inhand_Salary, Delay_from_due_date, Outstanding_Debt etc. as the values of the 75th percentile and maximum values varied by a huge margin, but since it is a financial dataset these could very well be valid data points, so we kept them for the analysis.<br>
Overall after the data wrangling process, no column had null or duplicated values and were consistent in nature, ready for analysis.
